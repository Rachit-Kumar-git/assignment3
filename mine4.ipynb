```python
name= input("Enter the student name")
cl= int(input("Enter the class"))
eng=float(input(f"Enter english marks of {name}"))
math=float(input(f"Enter the marks of maths {name}"))
hin=float(input(f"Enter the marks of hindi of {name}"))
sc=float(input(f"Enter science marks of  {name}"))
his=float(input(f"Enter the history marks of {name}"))
total=eng+math+hin+sc+his
percentage=(total/500)*100
print("Name: ", name)
print("Class:", cl)
print("Percentage %",percentage)
if(percentage>=60):
    print("Grade :'A'")
elif(percentage>=50 and percentage<60):
    print("Grade : 'B'")
elif(percentage>=40 and percentage<50 ):
    print("Grade: 'C'")
elif(percentage>=33 and percentage<40):
    print("Grade: 'D'")
else:
    print("Fail")
```

```python
number=int(input("Enter a number: "))
if(number>=2 and number<=50):
    num2=int(input("Enter the last multiplier number"))
    for i in range(1,num2+1):
        print(number,"*",i,"=",number*i)
else:
    print("ErroGiven number not in range!")
```

```python
user_input=input("Enter a number: ")
rev=user_input[::-1]
if(user_input==rev):
    print(user_input,"is a palindrome number!")
else:
    print(user_input,"is not a palindrome number! ")
```
